```python
import pandas as pd
import plotly.express as px

# Load data
df = pd.read_csv(
    r"C:\Users\Administrator\Desktop\data-viz-class-material\data\world_happiness_2023.csv"
)

df.head()
```

```python
# Top 20 countries by Generosity

top_generosity = (
    df.nlargest(20, 'Generosity')
      .sort_values('Generosity')
)

# Global average generosity
midpoint = df['Generosity'].mean()

fig = px.bar(
    top_generosity,
    x='Generosity',
    y='Country name',
    orientation='h',
    color='Generosity',
    color_continuous_scale='RdBu',
    color_continuous_midpoint=midpoint,
    title='Top 20 Countries by Generosity'
)

# Reference line
fig.add_vline(
    x=midpoint,
    line_dash='dash',
    annotation_text=f'Global Avg = {midpoint:.2f}'
)

fig.update_layout(
    template='simple_white',
    xaxis_title='Generosity Score',
    yaxis_title='Country'
)

fig.show()
```
